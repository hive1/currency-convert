```python
import requests
import json
import decimal

e_url = 'https://v6.exchangerate-api.com/v6/bc1b3aa3b38cbc50bb5f1457/latest/USD'

response = requests.get(e_url)
exchange_json = response.json()
```

```python
exchanges = {}
string_exchange = str(exchange_json['conversion_rates'])
list_se = string_exchange.split(',')
for value in list_se:
    ab, ex = value.split(":")

    ab = ab.replace("'","")
    print(ab)

    # Make sure that no {} get caught in the filter process
    ex = decimal.Decimal(ex.replace("}",""))
    print(ex)

    exchanges.update({ab[1:]: ex})

print(exchanges)
```

```python
def find(input):
    from csv import reader

    with open("currency_names.csv", "r") as file:
        reader = reader(file)
        for pair in reader:
            if pair[0].lower() == input.lower():
                
                # Returning full currencies
                return pair[1]
```

```python
amount = input("$: ")
tc = input("Abbreviation: ")
tc = tc.lower()
```

```python
for currency, conversions in exchanges.items():
    if currency.lower() == tc:
        converted_currency = Decimal(amount) * conversions
        converted_currency = round(converted_currency, 2)
        print(f"{currency}: {converted_currency}")
```
